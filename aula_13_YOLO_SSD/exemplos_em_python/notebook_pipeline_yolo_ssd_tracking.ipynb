# Pipeline sequencial: detecção e rastreamento
Este notebook organiza o fluxo completo com vídeo sintético, NMS, rastreamento por IoU, trilhas e métricas.

```python
from pathlib import Path
from vision_utils import create_synthetic_video
ROOT = Path.cwd()
video = create_synthetic_video(ROOT/'dados'/'notebook_video.avi', n_frames=120)
video
```

```python
from vision_utils import draw_scene, fake_detector, apply_nms, draw_detections
import cv2
frame = draw_scene(10)
dets = apply_nms(fake_detector(10), 0.25, 0.40)
out = draw_detections(frame, dets)
cv2.imwrite(str(ROOT/'saidas'/'notebook_deteccoes.png'), out)
```

```python
from vision_utils import IoUTracker
tracker = IoUTracker(iou_thr=0.30, trail_len=30)
for i in range(20):
    dets = apply_nms(fake_detector(i), 0.25, 0.40)
    tracked = tracker.update(dets)
print('IDs ativos:', [t['id'] for t in tracked])
```
